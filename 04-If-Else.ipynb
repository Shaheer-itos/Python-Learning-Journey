```python
#ask user for a number and display even or odd
num = int(input('Enter a number: '))
if num % 2 == 0:
  print('Even')
else:
  print('Odd')
```

```python
#ask user for two number and display the larger number
num1 = int(input('ENter first number: '))
num2 = int(input('ENter second number: '))
if num1 > num2:
  print(num1, 'is greater')
elif num1 == num2:
  print('Both numbers are same')
else:
  print(num2, 'is greater')
# elif: combination of else-if
#if/else is useful for two decisions
#if there are more than two outcomes,, we use elif
```

```python
#ask user for a number and display negative or positive
#use elif
num = int(input('ENter a number: '))
if num > 0:
  print('Positive')
elif num == 0:
  print('The number is 0')
else:
  print('Negative')
```

```python
#ask user for marks and print the grade
# 90+ → A
# 80-89 → B
# 70-79 → C
# 60-69 → D
# Below 60 → F
marks = float(input('Enter marks: '))
if marks >= 90:
  print('A Grade')
elif marks >= 80:
  print('B Grade')
elif marks >= 70:
  print('C Grade')
elif marks >= 60:
  print('D grade')
else:
  print('Fail')
```
