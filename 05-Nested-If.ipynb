```python
#ATM WIthdrawal
#ASk user:
#Enter account balance:
#Enter withdrawal amount
# Rules:
# If withdrawal amount is greater than balance:
# Insufficient balance
# otherwise caluclate remaining balance
# if remaining balance is below 1000:
# warning: low balance
# otherwise: transactation successful

balance = float(input('Enter account balance: '))
amount = float(input('Enter withdrawal amount: '))
if amount > balance:
  print('Insufficient balance')
elif balance - amount < 1000 :
  print('You have withdrawn Rs.',amount)
  print('Low balance')
else:
  print('You have withdrawn Rs.',amount)
```

```python
# A cinema has three rules:
# customer must be 18 or older
# if they are 18+, they must have a valid ticket
#if they have a ticket, they can enter
age = int(input('Enter age: '))
ticket = input('Do you have a ticket? ')
if age >= 18 and ticket == 'yes':
  print('Entry allowed')
else:
  print('Entry denied')
```

```python
'''
if they're under 18, tell they're underage
if they're 18 or older but don't have a ticket
tell them they need a ticked
Only allow entry of both conditions are satisfied

to solve this problem, we need to use nested-If
'''
age = int(input('Enter age: '))
if age >= 18:
  ticket = input('Do you have a ticket? ')
  if  ticket == 'yes':
    print('ENtry allowed')
  else:
    print('Buy a ticket')
else:
  print('You are underage')
```

```python
# a university student can enroll in a course if
# they have completed th prerequisite course
# and their fees is paid
prereq = input('Have you completed the course: ')
if prereq == 'yes':
  fee = input('Have you paid the fees: ')
  if fee == 'yes':
    print('Enrollment successful')
  else:
    print('Please pay your fees')
else:
  print('You must complete the prerequisite first')
```

```python
# Ask passenger age.
# If age is 18 or above, ask whether they have a valid passport.
# If they have a passport, ask whether they have a ticket.
# If they have both:
# Ask whether their luggage exceeds 30 kg.
# If yes → "Extra baggage fee required"
# Otherwise → "Check-in successful"
# If no passport → "Passport required"
# If no ticket → "Ticket required"
# If under 18 → "Age verification required"
age = int('Enter age: ')
if age >= 18:
  passport = input('Do you have a valid passport:')
  if passport == 'yes':
    ticket = input('Do you have a ticket: ')
    if ticket == 'yes':
      luggage = input('Do you have excess luggage: ')
      if luggage == 'yes':
        print('Extra baggage fees required')
      else:
        print('Check in successful')
    else:
      print('Ticket required')
  else:
    print('Passport required')
else:
  print('Age verification required')
```
