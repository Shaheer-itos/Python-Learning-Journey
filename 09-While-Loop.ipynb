```python
# ask user to enter a password. they have unlimited
# attempts, the program should stop
# as soon as they enter correct password
for i in range(3):
  password = input('Enter password: ')
  if password == 'python1234':
    print('Correct password')
```

```python
# While loop
i = 1
while i <= 5:
  print(i)
  i += 1
```

```python
# print numbers from 10 to 1 using while loop
i = 10
while i >= 1:
  print(i)
  i -= 1
```

```python
# print even numbers from 2 to 20 using while loop
i = 2
while i <= 20:
  print(i)
  i += 2
```

```python
i = 1
while i <= 20:
  if i % 2 == 0:
    print(i)
  i += 1
```

```python
# keep asking user for a number until they
# enter a negative number
number = 0
while number >= 0:
  number = int(input('ENter a number: '))
print('Negative number entered!!!')
```

```python
# sum numbers from 1 to 100 using while loop
total = 0
i = 1
while i <= 100:
  total += i
  i += 1
print(total)
```
