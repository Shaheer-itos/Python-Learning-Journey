```python
# # ask user for a password
# if password contains only letters print that
# if passeword contains only numbers print that
# if password contains both, print password accepted
passw = input('Enter password: ')
if passw.isalpha():
  print('you have entered only alphabets')
elif passw.isdigit():
  print('you have entered only digits')
elif passw.isalnum():
  print('Password accepted')
else:
  print('Wrong password')
```

```python
# # ask user for employee ID
# Must start with "EMP"
# Remaining part must contain digits
# Total length must be 8 characters
# sample employee ID: EMP12345
empid = input('Enter employee ID: ')
if empid.startswith('EMP') and len(empid) == 8 and empid[3:].isdigit():
  print('Valid ID')
else:
  print('Invalid ID')
```
