```python
def analyze_marks(marks):
  total = 0
  highest = marks[0]
  lowest = marks[0]

  for mark in marks:
    total += mark

    if mark > highest:
      highest = mark

    if mark < lowest:
      lowest = mark
  print(total)
  print(highest)
  print(lowest)
```

```python
marks = [75,88,62,91,70]
analyze_marks(marks)
```

```python
def calculate_bill(price, tax=5):
  tax_amount = price * tax / 100
  total = price + tax_amount
  return total
print(calculate_bill(1000))
print(calculate_bill(1000,10))
```

```python
def calculate_total(price,qty):
  return price * qty

def calculate_discount(amount):
  return amount * 0.10

def final_bill(price, qty):
  total = calculate_total(price,qty)
  discount = calculate_discount(total)
  return total - discount
```

```python
bill = final_bill(1000,2)
print('Final Bill:',bill)
```
