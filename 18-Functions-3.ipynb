```python
def analyze_marks(marks):
  total = 0
  highest = marks[0]
  lowest = marks[0]
  for mark in marks:
    total += mark
    if mark > highest:
      highest = mark
    if mark < lowest:
      lowest = mark
  return total, highest, lowest
```

```python
marks = [75,88,62,91,70]
t,h,l = analyze_marks(marks)
print(t)
print(h)
print(l)
```

```python
# create a fn analyze_numbers
# return total, number of even numbers, odd numbers
# numbers = [10, 15, 22, 31, 40, 51]
def analyze_number(numbers):
  total = 0
  even = 0
  odd = 0
  for number in numbers:
    total += number
    if number%2 == 0:
      even += 1
    else:
      odd += 1
  return total, even, odd
```

```python
numbers = [10,15,22,31,40,51]
tot, ev, od = analyze_number(numbers)
print(tot,ev,od)
```

```python
def get_high_marks(marks):
  high_marks = []
  for mark in marks:
    if mark>= 80:
      high_marks.append(mark)
  return high_marks
marks = [65,82,91,74,88,59]
result = get_high_marks(marks)
result
```

```python
# write a function double_numbers
# fn takes a list as input
# fn return the list in return with every number doubled
# numbers = [2,4,6,8]
#output = [4,8,12,16]
def double_numbers(numbers):
  doubled = []
  for i in numbers:
    doubled.append(i*2)
  return doubled
numbers = [2,3,5,6,7]
result = double_numbers(numbers)
result
```
