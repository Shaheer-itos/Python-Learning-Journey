```python
def analyze_marks(marks):
  total = 0
  highest = marks[0]
  lowest = marks[0]
  for mark in marks:
    total += mark
    if mark > highest:
      highest = mark
    if mark < lowest:
      lowest = mark
  return total, highest, lowest
```

```python
marks = [75,88,62,91,70]
total,highest,lowest = analyze_marks(marks)
print(total)
print(highest)
print(lowest)
```
