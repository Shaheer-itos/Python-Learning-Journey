```python
python_students = {'Ali','Sara','Ahmed','Zoya'}
ai_students = {'Sara','Zoya',"Hamza",'Bilal'}
# Union - combine both groups
all_students = ai_students | python_students
print(all_students)
# | (pipe) symbol is used for union in sets
all_students = python_students.union(ai_students)
all_students
```

```python
# Intersections: find shared members
both_courses = python_students & ai_students
print(both_courses)
# & (ampersand) sing is used for finding intersection
```

```python
both_courses = python_students.intersection(ai_students)
both_courses
```

```python
# Difference - find members is one group but not the other
python_only = python_students - ai_students
python_only
```

```python
# Symmetric difference - find members who are in exactly one group
one_course_only = python_students ^ ai_students
one_course_only
```

```python
a = {1,2,3,4}
b = {3,4,5,6}
print(a | b) #union (every item whithout duplication)
print(a & b) # interdection (only common items)
print(a - b) # difference (items in a but not in b)
print(a ^ b) # symm diff (only not common items)
```
