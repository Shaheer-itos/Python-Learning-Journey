```python
abc = open('subjects.txt','r')
subjects = abc.readlines() # create a list
print(subjects)
abc.close()
```

```python
# remove \n from the name of each subject
new_subjects = []
for subject in subjects:
  subject = subject.strip()
  new_subjects.append(subject)
new_subjects
```

```python
new = []
for s in subjects:
  new.append(s.replace('\n',''))
new
```
