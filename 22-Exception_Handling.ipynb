```python
# EXCEPTION HANDLING:
# exception handling lets us decide what our program
# should do when a particular runtime exception occurs
```

```python
number = int(input('Enter a whole number: '))
print('You entered:',number)
```

```python
try:
  number = int(input('Enter a whole number: '))
  print('You entered:',number)
except ValueError:
  print('This is not a valid whole number..')
```

```python
try:
  first = float(input('Enter first number: '))
  second = float(input('Enter second number: '))
  result = first / second
except ValueError:
  print('Please enter a valid number')
except ZeroDivisionError:
  print('The second number can not be zero')
else:
  print('Result:',result)
```

```python
try: # CHECKS THE CODE FOR EXCEPTION
  number = int(input('Enter a number: '))
except ValueError: # RUN IN CASE OF EXCEPTION
  print('Invalid number')
else: # RUNS IN CASE OF NO EXCEPTION
  print(number)
finally: # ALWAYS RUNS WHETHER THERE IS EXCEPTION OR NOT
  print('Input attempt complete')
```
