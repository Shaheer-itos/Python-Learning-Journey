```python
# List Comprehension
# a = [1,2,3,4] -> b=[1,4,9,16]
numbers = [1,2,3,4]
squares = []
for number in numbers:
  square = number * number
  squares.append(square)
print(squares)
```

```python
squares = [number * number for number in numbers]
squares
# new_list = [expression for item in existing_list]
```

```python
numbers = [2,3,4]
doubled = [number * 2 for number in numbers]
doubled
```

```python
prices = [100,200,300]
# add 50 to each item in price using list comp..
new_price = [price+50 for price in prices]
new_price
```

```python
names = ['ali','sara','ahmed']
# convert these names to uppercase using list comp..
capital = [n.upper() for n in names]
capital
```

```python
marks = [35,78,92,45,88]
# create a list of marks where marks >= 50
passed = [mark for mark in marks if mark >= 50]
passed
# new_list = [expression for item in old_list if condition]
```

```python
names = ['Ali','Ahmed','Sara','Fatima','Usman']
# add names to a new list that are longer than 4 characters
long_names = [name for name in names if len(name) > 4]
long_names
```

```python
# in the new_list, insert the mark if >50 else insert 0
passed = [mark if mark >= 50 else 0 for mark in marks]
# [VALUE_IF_TRUE if CONDITION else VALUE_IF_FALSE for ITEM in LIST]
passed
```

```python
names = ['ali','sara','ahmed']
# create a new list in which every name becomes uppercase
# if its length>3, otherwise keep the original name
new_names = [n.upper() if len(n)>3 else n for n in names]
new_names
```

```python
numbers = [1,2,3,4,5,6,7,8]
# list containing the word 'Even' / 'Odd' depending on the
# existing list
result = ['Even' if num%2==0 else 'Odd' for num in numbers]
result
```
