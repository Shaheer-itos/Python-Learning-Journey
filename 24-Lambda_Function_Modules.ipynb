```python
marks = [92,76,48,85,33]
# 85 or above: A
# 70 to 84: B
# 50 to 69: C
# Below 50: D
# write a list comprehension
grades = ['A' if m>=85 else 'B' if m>=70 else 'C'
if m>=50 else 'D' for m in marks]
grades
```

```python
num1 = [1,2,3]
num2 = [10,20]
result = [a*b for a in num1 for b in num2]
result
```

```python
result = [a*b for a in num1 for b in num2 if a*b > 20]
result
```

```python
# lambda function is a small function that we can
# write in a single expression
# also called anonymous function / nameless
```

```python
def double(num):
  return num*2
double(5)
```

```python
double = lambda num: num * 2
double(10)
# double is the variable holding the function
# lambda keyword introduces the function
# num is the parameter
# : separated the parameter from the expression
# num * 2 is the expression whose result is returned
```

```python
add_five = lambda num: num + 5
add_five(10)
```

```python
square = lambda num: num*num
square(7)
```
