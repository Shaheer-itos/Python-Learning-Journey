```python
# guessing game with user defined range and unlimited guesses
print('Welcome to the guessing game')
a = int(input('Enter a starting point: '))
b = int(input('Enter a ending point: '))
print(f'I am thinking of a number between {a} and {b}')
secret = random.randint(a,b)
while True:
  guess = int(input('Enter your guess: '))
  if secret == guess:
    print('You are right')
    break
  else:
    print('Wrong guess')
```

```python
# game with 3 guesses
# tell user about remaining chances after every wrong answer
print('Welcome to the guessing game')
a = int(input('Enter a starting point: '))
b = int(input('Enter a ending point: '))
print(f'I am thinking of a number between {a} and {b}')
secret = random.randint(a,b)
chances = 3
while chances>0:
  guess = int(input('Enter your guess: '))
  if secret == guess:
    print('You are right')
    break
  else:
    print('Wrong guess')
    chances -= 1
    if chances>0:
      print('Tries left:',chances)
    else:
      print('Game over!!! No more tries')
      print('The correct answer was',secret)
```

```python
# guessing game with user defined range,, 3 guesses, and a hint
print('Welcome to the guessing game')
a = int(input('Enter a starting point: '))
b = int(input('Enter a ending point: '))
print(f'I am thinking of a number between {a} and {b}')
secret = random.randint(a,b)
chances = 3
while chances>0:
  guess = int(input('Enter your guess: '))
  if secret == guess:
    print('You are right')
    break
  elif secret > guess:
    print('Your guess is too low')
    chances -= 1
    if chances>0:
      print('Tries left:',chances)
    else:
      print('Game over!!! No more tries')
      print('The correct answer was',secret)
  else:
    print('Your guess is too high')
    chances -= 1
    if chances>0:
      print('Tries left:',chances)
    else:
      print('Game over!!! No more tries')
      print('The correct answer was',secret)
```
